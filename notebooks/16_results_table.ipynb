```python
import os
import pandas as pd

files = {
    "Random": "results_random_baseline_dev.csv",
    "Dense, fixed 1000": "results_dense_fixed1000_dev.csv",
    "Dense, clause-aware 1500": "results_colab_clause1500_dev.csv",
    "Dense, keyword queries": "results_dense_keywords_clause1500_dev.csv",
    "BM25": "results_bm25_clause1500_dev.csv",
    "Hybrid 1:1": "results_hybrid_1_1_clause1500_dev.csv",
    "Hybrid 1:2": "results_hybrid_1_2_clause1500_dev.csv",
    "Hybrid keywords 1:1": "results_hybrid_kw_1_1_clause1500_dev.csv",
    "Hybrid 1:2 + rerank": "results_hybrid_1_2_rerank10_clause1500_dev.csv",
}

rows = {}
for name, f in files.items():
    path = f"../eval/{f}"
    if os.path.exists(path):
        df = pd.read_csv(path, index_col=0)
        rows[name] = df[["hit@1", "hit@3", "hit@10", "mrr"]].mean()
    else:
        print("missing:", f)

table = pd.DataFrame(rows).T.round(3)
table.to_csv("../eval/results_summary_dev.csv")
table
```
