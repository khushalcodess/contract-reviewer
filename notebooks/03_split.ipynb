```python
import sys, json
sys.path.append("..")
from src.data import load_cuad, TARGET_CLAUSES
from src.split import split_contracts, company_key

contracts = load_cuad()
dev_idx, test_idx = split_contracts(contracts)
print("dev:", len(dev_idx), "| test:", len(test_idx))

dev_co = {company_key(contracts[i]["title"]) for i in dev_idx}
test_co = {company_key(contracts[i]["title"]) for i in test_idx}
print("companies in both sets:", len(dev_co & test_co))

for name, idx in [("dev", dev_idx), ("test", test_idx)]:
    print(name, {cl: sum(1 for i in idx if contracts[i]["gold"][cl]) for cl in TARGET_CLAUSES})

with open("../eval/split.json", "w") as f:
    json.dump({
        "seed": 42,
        "dev": [contracts[i]["title"] for i in dev_idx],
        "test": [contracts[i]["title"] for i in test_idx],
    }, f, indent=2)
```
